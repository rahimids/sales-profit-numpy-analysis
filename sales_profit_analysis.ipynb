```python
q1=np.percentile(sales,25)
q3=np.percentile(sales,75)

iqr=q3-q1

print("Q1",q1)
print("Q3",q3)
print("IQR",iqr)
```

```python
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
print("Lower_Bound:",lower_bound)
print("Upper_Bound:",upper_bound)
```

```python
plt.figure(figsize=(8,5))
plt.hist(sales,bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(df["Sales"],df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

### Analysis

- Most orders have sales below 5,000.
- There is a general positive relationship between Sales and Profit.
- Orders with similar sales values can have very different profit values.
- Some high-sales orders generate losses, while others generate significant profits.
- This suggests that factors other than Sales also affect Profit.

```python
profit = df["Profit"].to_numpy()

correlation = np.corrcoef(sales, profit)[0,1]
print(correlation)
```

### Correlation Analysis

The correlation between Sales and Profit is 0.47, indicating a moderate positive relationship. Higher sales generally tend to generate higher profits, but sales alone cannot fully explain profit variations.
